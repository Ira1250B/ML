```python
#import basic libs
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#to handle missing values
df.isnull() #another way to look for null values and will return in boolean.
```

```python
df.isnull().sum()# as we need nuber of observations having null values
```

```python
#imputing the missing values -> unknown
#na means not available value
df['director']=df['director'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['cast']=df['cast'].fillna('Unknown')
df['country']=df['country'].fillna('Unknown')
df['date_added']=df['date_added'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['duration'].mode()[0] #outcome is series
```

```python
df['rating']=df['rating'].fillna(df['rating'].mode()[0])
df['duration']=df['duration'].fillna(df['duration'].mode()[0])
```

```python
df.isnull().sum()
```
