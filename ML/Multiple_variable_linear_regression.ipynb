```python
def gradient_descent(theta_array, x1, x2, x3, x4, y, m, alpha):

    sum_0, sum_1, sum_2, sum_3, sum_4 = 0, 0, 0, 0, 0

    for i in range(m):

        error = (
            theta_array[0]
            + theta_array[1] * x1[i]
            + theta_array[2] * x2[i]
            + theta_array[3] * x3[i]
            + theta_array[4] * x4[i]
            - y[i]
        )

        sum_0 += error
        sum_1 += x1[i] * error
        sum_2 += x2[i] * error
        sum_3 += x3[i] * error
        sum_4 += x4[i] * error


    update_theta0 = theta_array[0] - (alpha/m) * sum_0
    update_theta1 = theta_array[1] - (alpha/m) * sum_1
    update_theta2 = theta_array[2] - (alpha/m) * sum_2
    update_theta3 = theta_array[3] - (alpha/m) * sum_3
    update_theta4 = theta_array[4] - (alpha/m) * sum_4


    new_theta = [
        update_theta0,
        update_theta1,
        update_theta2,
        update_theta3,
        update_theta4
    ]

    return new_theta
```
