```python
#polynomial regression
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
#for trainning and analysing how my model is behaving batter
from sklearn.model_selection import train_test_split
#standerd scaler is for normalization of all features into one range
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
data=load_diabetes()
```

```python
data
```

```python
x=pd.DataFrame(data.data,columns=data.feature_names)#data.data is for the name data in above dataset and columns we are considering features
y=data.target
```

```python
x
```

```python
#extracting sub features
features=['bmi','bp','s5']
X=x[features]
```

```python
X
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.2)
```

```python
scaler=StandardScaler()#feature normalization
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
def costfunction(y_actual,y_predict):
  m=len(y_actual)
  summ= np.sum((y_predict-y_actual)**2)
  return (summ/(2*m))
```

```python
#I need to find which degree of ploynomial is best for analysis hence now wewill perform actual regression
degree=range(1,7)#we can take any numbers here
cost_values=[]# storing cost values for each parameter
for deg in degree:
  poly=PolynomialFeatures(degree=deg,include_bias=False)# to check original effect of feature on each other we are not including bias it is for analysis only
  x_train_poly=poly.fit_transform(X_train)#we are doing this to check effect of each cost function on each parameter
  x_test_poly=poly.transform(X_test)
  model = LinearRegression()
  model.fit(X_train, Y_train)
  Y_predict = model.predict(X_test)
```

```python
plt.scatter(Y_test, Y_predict)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Multiple Linear Regression")
plt.show()
```
